# Number System Converter

One of my very first programming projects, written back in high school.

We had just learned about the ASCII table and how computers represent numbers in different bases, and I got curious about how a conversion actually works under the hood. So instead of using Python's built-in `bin()` or `hex()`, I decided to build my own converter from scratch and implement the math by hand.

It's an interactive command-line program: you pick a conversion from the menu (decimal, binary, hexadecimal, octal) and it converts your number, with input checking so you can't type invalid digits.

I'm keeping the original code as I wrote it back then. It's not the prettiest, but it works and it's honestly mine.
